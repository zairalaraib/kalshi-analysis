```python
import requests
import pandas as pd

BASE_URL = "https://external-api.kalshi.com/trade-api/v2"

url = f"{BASE_URL}/markets"

params = {
    "status": "open",
    "limit": 100
}

response = requests.get(url, params=params)
response.raise_for_status()

data = response.json()

# Turn the markets into a dataframe
markets = pd.DataFrame(data["markets"])

print("Number of markets returned:", len(markets))

print("\nALL COLUMNS:")
print(markets.columns.tolist())

print("\nFIRST 5 MARKETS:")
display(markets.head())

print("\nCURSOR:")
print(data.get("cursor"))
```

```python
market_view = markets[
    [
        "ticker",
        "event_ticker",
        "title",
        "status",
        "last_price_dollars",
        "yes_bid_dollars",
        "yes_ask_dollars",
        "yes_bid_size_fp",
        "yes_ask_size_fp",
        "volume_fp",
        "volume_24h_fp",
        "liquidity_dollars",
        "open_interest_fp"
    ]
].copy()

numeric_cols = [
    "last_price_dollars",
    "yes_bid_dollars",
    "yes_ask_dollars",
    "yes_bid_size_fp",
    "yes_ask_size_fp",
    "volume_fp",
    "volume_24h_fp",
    "liquidity_dollars",
    "open_interest_fp"
]

for col in numeric_cols:
    market_view[col] = pd.to_numeric(market_view[col], errors="coerce")

market_view = market_view.sort_values(
    "liquidity_dollars",
    ascending=False
)

display(market_view.head(20))
```

```python
orderbook_url = f"{BASE_URL}/markets/{ticker}/orderbook"

response = requests.get(orderbook_url)
response.raise_for_status()

orderbook_data = response.json()

print(orderbook_data)
```

```python
book = orderbook_data["orderbook_fp"]

# YES bids
yes_book = pd.DataFrame(
    book["yes_dollars"],
    columns=["price", "quantity"]
)

# NO bids
no_book = pd.DataFrame(
    book["no_dollars"],
    columns=["price", "quantity"]
)

# Convert strings → numbers
yes_book["price"] = yes_book["price"].astype(float)
yes_book["quantity"] = yes_book["quantity"].astype(float)

no_book["price"] = no_book["price"].astype(float)
no_book["quantity"] = no_book["quantity"].astype(float)

print("YES ORDER BOOK")
display(yes_book.sort_values("price", ascending=False))

print("\nNO ORDER BOOK")
display(no_book.sort_values("price", ascending=False))
```

```python
# YES BIDS
yes_bids = yes_book.copy()
yes_bids.columns = ["bid_price", "bid_quantity"]

# YES ASKS are implied by NO bids
yes_asks = no_book.copy()
yes_asks["price"] = 1 - yes_asks["price"]
yes_asks.columns = ["ask_price", "ask_quantity"]


# NO BIDS
no_bids = no_book.copy()
no_bids.columns = ["bid_price", "bid_quantity"]

# NO ASKS are implied by YES bids
no_asks = yes_book.copy()
no_asks["price"] = 1 - no_asks["price"]
no_asks.columns = ["ask_price", "ask_quantity"]


# Sort like a normal order book
yes_bids = yes_bids.sort_values("bid_price", ascending=False)
yes_asks = yes_asks.sort_values("ask_price", ascending=True)

no_bids = no_bids.sort_values("bid_price", ascending=False)
no_asks = no_asks.sort_values("ask_price", ascending=True)


print("YES BIDS")
display(yes_bids.head(10))

print("\nYES ASKS")
display(yes_asks.head(10))

print("\nNO BIDS")
display(no_bids.head(10))

print("\nNO ASKS")
display(no_asks.head(10))
```

```python
best_yes_bid = yes_bids["bid_price"].max()
best_yes_ask = yes_asks["ask_price"].min()

spread = best_yes_ask - best_yes_bid
midprice = (best_yes_bid + best_yes_ask) / 2

print("Best YES bid:", best_yes_bid)
print("Best YES ask:", best_yes_ask)
print("Spread:", spread)
print("Midprice:", midprice)
```

```python
best_no_bid = no_bids["bid_price"].max()
best_no_ask = no_asks["ask_price"].min()

spread = best_no_ask - best_no_bid
midprice = (best_no_bid + best_no_ask) / 2

print("Best NO bid:", best_no_bid)
print("Best NO ask:", best_no_ask)
print("Spread:", spread)
print("Midprice:", midprice)
```

```python
# Look at the top 5 price levels on each side

top_yes_bids = yes_bids.head(5)
top_yes_asks = yes_asks.head(5)

bid_depth = top_yes_bids["bid_quantity"].sum()
ask_depth = top_yes_asks["ask_quantity"].sum()

imbalance = (bid_depth - ask_depth) / (bid_depth + ask_depth)

print("Top 5 bid depth:", bid_depth)
print("Top 5 ask depth:", ask_depth)
print("Order book imbalance:", imbalance)
```

```python
# Prepare bids from best → worse
bid_depth = yes_bids.sort_values("bid_price", ascending=False).copy()
bid_depth["cumulative_quantity"] = bid_depth["bid_quantity"].cumsum()

# Prepare asks from best → worse
ask_depth = yes_asks.sort_values("ask_price", ascending=True).copy()
ask_depth["cumulative_quantity"] = ask_depth["ask_quantity"].cumsum()

plt.figure(figsize=(10, 5))

plt.step(
    bid_depth["bid_price"],
    bid_depth["cumulative_quantity"],
    where="post",
    label="YES bids"
)

plt.step(
    ask_depth["ask_price"],
    ask_depth["cumulative_quantity"],
    where="post",
    label="YES asks"
)

plt.xlabel("YES Price")
plt.ylabel("Cumulative Quantity")
plt.title("YES Cumulative Order Book Depth")
plt.legend()

plt.show()
```

```python
# Prepare NO bids from best → worse
no_bid_depth = no_bids.sort_values("bid_price", ascending=False).copy()
no_bid_depth["cumulative_quantity"] = no_bid_depth["bid_quantity"].cumsum()

# Prepare NO asks from best → worse
no_ask_depth = no_asks.sort_values("ask_price", ascending=True).copy()
no_ask_depth["cumulative_quantity"] = no_ask_depth["ask_quantity"].cumsum()

plt.figure(figsize=(10, 5))

plt.step(
    no_bid_depth["bid_price"],
    no_bid_depth["cumulative_quantity"],
    where="post",
    label="NO bids"
)

plt.step(
    no_ask_depth["ask_price"],
    no_ask_depth["cumulative_quantity"],
    where="post",
    label="NO asks"
)

plt.xlabel("NO Price")
plt.ylabel("Cumulative Quantity")
plt.title("NO Cumulative Order Book Depth")
plt.legend()

plt.show()
```

```python
# Best YES prices
best_bid = yes_bids["bid_price"].max()
best_ask = yes_asks["ask_price"].min()

# Quantity at best prices
best_bid_qty = yes_bids.loc[
    yes_bids["bid_price"] == best_bid,
    "bid_quantity"
].sum()

best_ask_qty = yes_asks.loc[
    yes_asks["ask_price"] == best_ask,
    "ask_quantity"
].sum()

midprice = (best_bid + best_ask) / 2

microprice = (
    best_ask * best_bid_qty +
    best_bid * best_ask_qty
) / (best_bid_qty + best_ask_qty)

signal = microprice - midprice

print("Best bid:", best_bid)
print("Best ask:", best_ask)

print("Best bid quantity:", best_bid_qty)
print("Best ask quantity:", best_ask_qty)

print("Midprice:", round(midprice, 4))
print("Microprice:", round(microprice, 4))
print("Signal:", round(signal, 4))
```

```python
for levels in [1, 3, 5, 10]:

    bid_depth = yes_bids.head(levels)["bid_quantity"].sum()
    ask_depth = yes_asks.head(levels)["ask_quantity"].sum()

    imbalance = (
        (bid_depth - ask_depth) /
        (bid_depth + ask_depth)
    )

    print(
        f"Top {levels} levels imbalance:",
        round(imbalance, 3)
    )
```

```python
# Look near current YES price
near_bids = yes_bids.head(10).copy()
near_asks = yes_asks.head(10).copy()

# Median quantity
bid_median = near_bids["bid_quantity"].median()
ask_median = near_asks["ask_quantity"].median()

# Wall = at least 2x median size
bid_walls = near_bids[
    near_bids["bid_quantity"] >= 2 * bid_median
]

ask_walls = near_asks[
    near_asks["ask_quantity"] >= 2 * ask_median
]

print("BIG BID WALLS")
display(bid_walls)

print("\nBIG ASK WALLS")
display(ask_walls)
```

```python
from datetime import datetime
import pandas as pd

snapshot = {
    "timestamp": datetime.utcnow(),

    "midprice": midprice,
    "microprice": microprice,

    "microprice_edge": microprice - midprice,

    "spread": best_ask - best_bid,

    "best_bid": best_bid,
    "best_ask": best_ask,

    "best_bid_qty": best_bid_qty,
    "best_ask_qty": best_ask_qty
}

snapshot
```

```python
import time
from datetime import datetime, timezone

snapshots = []

for i in range(60):

    # Get fresh order book
    response = requests.get(
        f"{BASE_URL}/markets/{ticker}/orderbook"
    )
    response.raise_for_status()

    book = response.json()["orderbook_fp"]

    # Build fresh YES bid/ask books
    yes_temp = pd.DataFrame(
        book["yes_dollars"],
        columns=["price", "quantity"]
    )

    no_temp = pd.DataFrame(
        book["no_dollars"],
        columns=["price", "quantity"]
    )

    yes_temp = yes_temp.astype(float)
    no_temp = no_temp.astype(float)

    # Best YES bid
    best_bid = yes_temp["price"].max()

    # YES asks implied from NO bids
    no_temp["yes_ask_price"] = 1 - no_temp["price"]
    best_ask = no_temp["yes_ask_price"].min()

    # quantities at best levels
    best_bid_qty = yes_temp.loc[
        yes_temp["price"] == best_bid,
        "quantity"
    ].sum()

    best_ask_qty = no_temp.loc[
        no_temp["yes_ask_price"] == best_ask,
        "quantity"
    ].sum()

    # Mid + microprice
    midprice = (best_bid + best_ask) / 2

    microprice = (
        best_ask * best_bid_qty +
        best_bid * best_ask_qty
    ) / (best_bid_qty + best_ask_qty)

    snapshots.append({
        "timestamp": datetime.now(timezone.utc),
        "midprice": midprice,
        "microprice": microprice,
        "microprice_edge": microprice - midprice,
        "spread": best_ask - best_bid,
        "best_bid": best_bid,
        "best_ask": best_ask,
        "best_bid_qty": best_bid_qty,
        "best_ask_qty": best_ask_qty
    })

    print(
        i + 1,
        "mid:", round(midprice, 4),
        "micro:", round(microprice, 4),
        "edge:", round(microprice - midprice, 4)
    )

    time.sleep(10)

snapshots_df = pd.DataFrame(snapshots)

display(snapshots_df.head())
```

```python
start_mid = snapshots_df["midprice"].iloc[0]
end_mid = snapshots_df["midprice"].iloc[-1]

price_change = end_mid - start_mid

avg_edge = snapshots_df["microprice_edge"].mean()

print("Starting midpoint:", round(start_mid, 4))
print("Ending midpoint:", round(end_mid, 4))

print("Price change:", round(price_change, 4))

print("Average microprice edge:", round(avg_edge, 4))

if avg_edge < 0:
    print("Signal predicted: DOWN")
elif avg_edge > 0:
    print("Signal predicted: UP")
else:
    print("Signal predicted: NEUTRAL")

if price_change < 0:
    print("Actual move: DOWN")
elif price_change > 0:
    print("Actual move: UP")
else:
    print("Actual move: NO MOVE")
```

```python
snapshots_df["top_imbalance"] = (
    snapshots_df["best_bid_qty"] - snapshots_df["best_ask_qty"]
) / (
    snapshots_df["best_bid_qty"] + snapshots_df["best_ask_qty"]
)

print("Average imbalance:",
      round(snapshots_df["top_imbalance"].mean(), 3))

print("Minimum imbalance:",
      round(snapshots_df["top_imbalance"].min(), 3))

print("Maximum imbalance:",
      round(snapshots_df["top_imbalance"].max(), 3))

display(
    snapshots_df[
        ["timestamp", "midprice", "microprice", "top_imbalance"]
    ].head(10)
)
```

```python
# Normalize microprice edge by spread
snapshots_df["microprice_signal"] = (
    snapshots_df["microprice_edge"] /
    snapshots_df["spread"]
)

# Simple directional signal
snapshots_df["signal"] = 0

snapshots_df.loc[
    (snapshots_df["microprice_signal"] > 0.20) &
    (snapshots_df["top_imbalance"] > 0.30),
    "signal"
] = 1

snapshots_df.loc[
    (snapshots_df["microprice_signal"] < -0.20) &
    (snapshots_df["top_imbalance"] < -0.30),
    "signal"
] = -1

print("BUY signals:",
      (snapshots_df["signal"] == 1).sum())

print("SELL signals:",
      (snapshots_df["signal"] == -1).sum())

print("NO TRADE:",
      (snapshots_df["signal"] == 0).sum())

display(
    snapshots_df[
        [
            "timestamp",
            "midprice",
            "microprice_signal",
            "top_imbalance",
            "signal"
        ]
    ].head(20)
)
```

```python
flow_df = snapshots_df.copy()

flow_df["bid_qty_change"] = flow_df["best_bid_qty"].diff()
flow_df["ask_qty_change"] = flow_df["best_ask_qty"].diff()

flow_df["flow_signal"] = (
    flow_df["bid_qty_change"] -
    flow_df["ask_qty_change"]
)

display(
    flow_df[
        [
            "timestamp",
            "best_bid_qty",
            "best_ask_qty",
            "bid_qty_change",
            "ask_qty_change",
            "flow_signal"
        ]
    ].head(20)
)
```
